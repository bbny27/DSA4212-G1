```python
# Numerical calculations, plotting and timing.
import numpy as np
import matplotlib.pyplot as plt
import time

# Bayesian optimisation.
from skopt import gp_minimize
from skopt.space import Real

# Optional custom GP and kernels for BO.
from skopt.learning import GaussianProcessRegressor
from skopt.learning.gaussian_process.kernels import (
    ConstantKernel,
    Matern,
    RBF,
)

# CMA-ES.
import cma

# TuRBO-1 and its random-seed control.
import torch
from turbo import Turbo1
```

```python
def two_basin_function(x, c, a, b, r0):
    """Evaluate the two-basin objective at one point x."""
    x = np.asarray(x, dtype=float)
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    # Distance from each basin's centre.
    distance_a = np.linalg.norm(x - a)
    distance_b = np.linalg.norm(x - b)

    # Worse bowl: minimum value 1 at a.
    basin_a = 1.0 + distance_a**2

    # Better bowl: minimum value 0 at b.
    # The penalty is zero within radius r0 of b.
    penalty = c * max(0.0, distance_b - r0)**2
    basin_b = distance_b**2 + penalty

    # Combine the two bowls by taking the lower value.
    return float(min(basin_a, basin_b))
```

```python
class BudgetReached(Exception):
    pass


def limit_evaluations(objective, budget):
    """Return a counted objective and its evaluation history."""
    history = {"X": [], "y": []}

    def evaluate(x):
        if len(history["y"]) >= budget:
            raise BudgetReached

        value = float(objective(x))
        history["X"].append(np.asarray(x).copy())
        history["y"].append(value)
        return value

    return evaluate, history
```

```python
a = np.array([-2.0, 0.0])
b = np.array([2.0, 0.0])
r0 = 0.5
c = 3.0

bounds = np.array([[-4.0, 4.0], [-4.0, 4.0]])
lower, upper = bounds[:, 0], bounds[:, 1]

budget = 50
seed = 42
n_init = 4

def objective(x):
    return two_basin_function(x, c, a, b, r0)

results = {}
```

```python
limited_objective, history = limit_evaluations(objective, budget)
start = time.perf_counter()

try:
    bo_result = gp_minimize(
        func=limited_objective,
        dimensions=[Real(float(lo), float(hi)) for lo, hi in bounds],
        n_calls=budget,
        n_initial_points=n_init,
        acq_func="EI",
        noise=1e-10,
        random_state=seed,
    )
except BudgetReached:
    pass

results["BO"] = {
    "X": np.asarray(history["X"]),
    "y": np.asarray(history["y"]),
    "seconds": time.perf_counter() - start,
}
print("BO finished")
```

```python
limited_objective, history = limit_evaluations(objective, budget)
rng = np.random.default_rng(seed)

# CMA-ES searches in [0,1]^2; map its points to our original domain.
initial_mean = rng.uniform(size=2)

def cma_objective(u):
    x = lower + (upper - lower) * np.asarray(u)
    return limited_objective(x)

start = time.perf_counter()

try:
    cma_result = cma.fmin2(
        cma_objective,
        initial_mean,
        0.25,
        options={
            "bounds": [0.0, 1.0],
            "maxfevals": budget,
            "seed": seed,
            "verbose": -9,
            "verb_log": 0,
        },
    )
except BudgetReached:
    pass

results["CMA-ES"] = {
    "X": np.asarray(history["X"]),
    "y": np.asarray(history["y"]),
    "seconds": time.perf_counter() - start,
}
print("CMA-ES finished")
```

```python
limited_objective, history = limit_evaluations(objective, budget)

np.random.seed(seed)
torch.manual_seed(seed)

start = time.perf_counter()

try:
    turbo_result = Turbo1(
        f=limited_objective,
        lb=lower,
        ub=upper,
        n_init=n_init,
        max_evals=budget,
        batch_size=1,
        use_ard=True,
        n_training_steps=50,
        verbose=False,
        device="cpu",
        dtype="float64",
    )
    turbo_result.optimize()

except BudgetReached:
    pass

results["TuRBO-1"] = {
    "X": np.asarray(history["X"]),
    "y": np.asarray(history["y"]),
    "seconds": time.perf_counter() - start,
}
print("TuRBO-1 finished")
```

```python
for name, result in results.items():
    y = result["y"]
    hits = np.flatnonzero(y < 1.0)
    first_hit = int(hits[0] + 1) if len(hits) else None

    print(
        f"{name:8s} | evaluations={len(y):2d} "
        f"| best={y.min():.6f} "
        f"| first f<1: {first_hit} "
        f"| time={result['seconds']:.1f}s"
    )

    plt.plot(
        np.arange(1, len(y) + 1),
        np.minimum.accumulate(y),
        label=name,
    )

plt.axhline(1.0, color="black", linestyle="--", label="Worse minimum")
plt.xlabel("Objective evaluations")
plt.ylabel("Best function value so far")
plt.title("Experiment 1: c = 3, seed = 42")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
# REPEATED EXPERIMENT: c = 3, five seeds, 50 evaluations per method.
# First run cells 0-4 (installation, imports, objective, counter, settings).
# The original seed-42 results above are kept separately in results.
repeat_seeds = [0, 1, 2, 3, 4]
repeat_c = 3.0
repeat_budget = 50
repeat_n_init = 4
methods = ["BO", "CMA-ES", "TuRBO-1"]
repeated_results = {name: {} for name in methods}

def repeat_objective(x):
    return two_basin_function(x, repeat_c, a, b, r0)

for run_seed in repeat_seeds:
    for name in methods:
        limited_objective, history = limit_evaluations(
            repeat_objective, repeat_budget
        )
        start = time.perf_counter()
        try:
            if name == "BO":
                gp_minimize(
                    func=limited_objective,
                    dimensions=[Real(float(lo), float(hi)) for lo, hi in bounds],
                    n_calls=repeat_budget,
                    n_initial_points=repeat_n_init,
                    acq_func="EI", noise=1e-10,
                    random_state=run_seed,
                )
            elif name == "CMA-ES":
                # Use normalised coordinates, as in the single-run cell.
                rng = np.random.default_rng(run_seed)
                initial_mean = rng.uniform(size=len(bounds))
                cma.fmin2(
                    lambda u: limited_objective(
                        lower + (upper - lower) * np.asarray(u)
                    ),
                    initial_mean, 0.25,
                    options={
                        "bounds": [0.0, 1.0],
                        "maxfevals": repeat_budget,
                        # cma treats seed=0 specially; use a positive seed.
                        "seed": run_seed + 1,
                        "verbose": -9, "verb_log": 0,
                    },
                )
            else:
                np.random.seed(run_seed)
                torch.manual_seed(run_seed)
                optimiser = Turbo1(
                    f=limited_objective, lb=lower, ub=upper,
                    n_init=repeat_n_init, max_evals=repeat_budget,
                    batch_size=1, use_ard=True, n_training_steps=50,
                    verbose=False, device="cpu", dtype="float64",
                )
                optimiser.optimize()
        except BudgetReached:
            pass  # Stop before exceeding the common evaluation budget.

        y = np.asarray(history["y"], dtype=float)
        hits = np.flatnonzero(y < 1.0)
        repeated_results[name][run_seed] = {
            "X": np.asarray(history["X"]),
            "y": y,
            "best_so_far": np.minimum.accumulate(y),
            "first_hit": int(hits[0] + 1) if len(hits) else None,
            "seconds": time.perf_counter() - start,
        }
        print(
            f"seed={run_seed} | {name:8s} | evaluations={len(y)} "
            f"| best={y.min():.6f} | success={bool(len(hits))}",
            flush=True,
        )

print("All 15 runs completed. Run the next cell for the summary.")
```

```python
# SUMMARY: all five final values, success rates and median progress.
import pandas as pd
from IPython.display import display

if not all(len(repeated_results[name]) == len(repeat_seeds) for name in methods):
    raise RuntimeError("Run the repeated-experiment cell to completion first.")

rows = []
summary_rows = []
fig, ax = plt.subplots(figsize=(8, 5))

for name in methods:
    runs = [repeated_results[name][s] for s in repeat_seeds]
    final_values = [r["best_so_far"][-1] for r in runs]
    successes = sum(r["first_hit"] is not None for r in runs)
    for s, r in zip(repeat_seeds, runs):
        rows.append({
            "method": name, "seed": s, "evaluations": len(r["y"]),
            "final_best": r["best_so_far"][-1],
            "first_f_below_1": r["first_hit"],
            "seconds": r["seconds"],
        })
    summary_rows.append({
        "method": name,
        "successes": f"{successes}/{len(runs)}",
        "success_rate": successes / len(runs),
        "median_final_best": np.median(final_values),
        "median_seconds": np.median([r["seconds"] for r in runs]),
    })

    # If an optimiser stops early, carry its last best value forward.
    curves = np.stack([
        np.pad(r["best_so_far"],
               (0, repeat_budget - len(r["y"])), mode="edge")
        for r in runs
    ])
    median = np.median(curves, axis=0)
    q25, q75 = np.quantile(curves, [0.25, 0.75], axis=0)
    line, = ax.plot(np.arange(1, repeat_budget + 1), median, label=name)
    ax.fill_between(np.arange(1, repeat_budget + 1), q25, q75,
                    color=line.get_color(), alpha=0.15)

run_table = pd.DataFrame(rows)
summary_table = pd.DataFrame(summary_rows)
display(run_table)
display(summary_table)

ax.axhline(1.0, color="black", linestyle="--", label="Worse minimum")
ax.set_yscale("symlog", linthresh=1e-4)
ax.set_xlabel("Objective evaluations")
ax.set_ylabel("Best function value (median; shaded 25-75%)")
ax.set_title(f"Experiment 1: c = {repeat_c:g}, five seeds")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import json
from pathlib import Path

# Fixed conditions: only c and the random seed change.
c_values = [0.0, 3.0, 10.0, 30.0]
seeds = list(range(20))
methods = ["BO", "CMA-ES", "TuRBO-1", "Random"]

a = np.array([-2.0, 0.0])
b = np.array([2.0, 0.0])
r0 = 0.5
bounds = np.array([[-4.0, 4.0], [-4.0, 4.0]])
lower, upper = bounds[:, 0], bounds[:, 1]

budget = 50
n_init = 4

# Checkpoint configuration prevents accidentally mixing different setups.
config = {
    "c_values": c_values,
    "seeds": seeds,
    "methods": methods,
    "a": a.tolist(),
    "b": b.tolist(),
    "r0": r0,
    "bounds": bounds.tolist(),
    "budget": budget,
    "n_init": n_init,
    "bo_acquisition": "EI",
    "bo_noise": 1e-10,
    "cma_sigma0": 0.25,
    "turbo_training_steps": 50,
}

checkpoint = Path("experiment1_sweep.json")

if checkpoint.exists():
    saved = json.loads(checkpoint.read_text())
    if saved["config"] != config:
        raise ValueError(
            "Checkpoint settings differ. Rename the checkpoint "
            "before starting a different experiment."
        )
    sweep_runs = saved["runs"]
else:
    sweep_runs = []

completed = {
    (r["c"], r["seed"], r["method"])
    for r in sweep_runs
}
print(f"Resuming with {len(completed)}/320 trials completed.", flush=True)

for c_value in c_values:
    for run_seed in seeds:
        for name in methods:
            key = (c_value, run_seed, name)
            if key in completed:
                continue

            # Fix the objective for this difficulty.
            def objective(x):
                return two_basin_function(x, c_value, a, b, r0)

            # Fresh evaluation counter for every trial.
            evaluate, history = limit_evaluations(objective, budget)
            start = time.perf_counter()

            try:
                if name == "BO":
                    gp_minimize(
                        func=evaluate,
                        dimensions=[
                            Real(float(lo), float(hi))
                            for lo, hi in bounds
                        ],
                        n_calls=budget,
                        n_initial_points=n_init,
                        initial_point_generator="random",
                        acq_func="EI",
                        noise=1e-10,
                        random_state=run_seed,
                    )

                elif name == "CMA-ES":
                    rng = np.random.default_rng(run_seed)
                    initial_mean = rng.uniform(size=len(bounds))

                    # Search in [0,1]^2 and map back to the true domain.
                    cma.fmin2(
                        lambda u: evaluate(
                            lower + (upper - lower) * np.asarray(u)
                        ),
                        initial_mean,
                        0.25,
                        options={
                            "bounds": [0.0, 1.0],
                            "maxfevals": budget,
                            # Avoid cma's special treatment of seed=0.
                            "seed": run_seed + 1,
                            "verbose": -9,
                            "verb_log": 0,
                        },
                    )

                elif name == "TuRBO-1":
                    np.random.seed(run_seed)
                    torch.manual_seed(run_seed)

                    optimiser = Turbo1(
                        f=evaluate,
                        lb=lower,
                        ub=upper,
                        n_init=n_init,
                        max_evals=budget,
                        batch_size=1,
                        use_ard=True,
                        n_training_steps=50,
                        verbose=False,
                        device="cpu",
                        dtype="float64",
                    )
                    optimiser.optimize()

                else:
                    # Independent uniform random-search baseline.
                    rng = np.random.default_rng(run_seed)
                    points = rng.uniform(
                        lower, upper,
                        size=(budget, len(bounds)),
                    )
                    for x in points:
                        evaluate(x)

            except BudgetReached:
                pass

            elapsed = time.perf_counter() - start
            y = np.asarray(history["y"], dtype=float)
            hits = np.flatnonzero(y < 1.0)

            record = {
                "c": c_value,
                "seed": run_seed,
                "method": name,
                "evaluations": len(y),
                "success": bool(len(hits)),
                "first_hit": int(hits[0] + 1) if len(hits) else None,
                "final_best": float(y.min()),
                "seconds": elapsed,
                "X": np.asarray(history["X"]).tolist(),
                "y": y.tolist(),
            }
            sweep_runs.append(record)
            completed.add(key)

            # Save after every completed trial.
            temporary = checkpoint.with_suffix(".tmp")
            temporary.write_text(json.dumps({
                "config": config,
                "runs": sweep_runs,
            }))
            temporary.replace(checkpoint)

            print(
                f"[{len(completed):3d}/320] c={c_value:g} "
                f"| seed={run_seed:2d} | {name:8s} "
                f"| n={len(y)} | best={y.min():.6f} "
                f"| success={bool(len(hits))} | {elapsed:.1f}s",
                flush=True,
            )

print("Experiment complete.")
```

```python
import pandas as pd
from IPython.display import display

# Reload saved results, so this also works after reconnecting.
saved = json.loads(Path("experiment1_sweep.json").read_text())
sweep_runs = saved["runs"]
cfg = saved["config"]
methods = cfg["methods"]
c_values = cfg["c_values"]
budget = cfg["budget"]

# Individual trial results.
trial_table = pd.DataFrame([
    {k: v for k, v in r.items() if k not in ("X", "y")}
    for r in sweep_runs
])

# Summary for each difficulty and method.
summary = (
    trial_table.groupby(["c", "method"], sort=False)
    .agg(
        runs=("success", "size"),
        successes=("success", "sum"),
        success_rate=("success", "mean"),
        median_final_best=("final_best", "median"),
        median_seconds=("seconds", "median"),
        min_evaluations=("evaluations", "min"),
        max_evaluations=("evaluations", "max"),
    )
    .reset_index()
)

display(summary)
trial_table.to_csv("experiment1_trials.csv", index=False)
summary.to_csv("experiment1_summary.csv", index=False)

# Keep method colours consistent across all plots.
colours = dict(zip(methods, ["tab:blue", "tab:orange", "tab:green", "tab:red"]))

# Plot 1: probability of beating the worse minimum within the budget.
fig, ax = plt.subplots(figsize=(7, 4))

for name in methods:
    part = summary[summary["method"] == name].sort_values("c")
    ax.plot(
        part["c"], part["success_rate"],
        marker="o", label=name, color=colours[name],
    )

ax.set_xscale("symlog", linthresh=1)
ax.set_xticks(c_values)
ax.set_xticklabels([f"{c:g}" for c in c_values])
ax.set_ylim(-0.05, 1.05)
ax.set_xlabel("Penalty strength c")
ax.set_ylabel(f"Success rate: observed f < 1 within {budget} evaluations")
ax.set_title("Discovery of a solution better than the worse minimum")
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()

# Plot 2: progress curves for each difficulty.
fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True, sharey=True)
evaluation_counts = np.arange(1, budget + 1)

for ax, c_value in zip(axes.flat, c_values):
    for name in methods:
        runs = [
            r for r in sweep_runs
            if r["c"] == c_value and r["method"] == name
        ]
        if not runs:
            continue

        # Retain the last best value if a method stopped early.
        curves = []
        for r in runs:
            curve = np.minimum.accumulate(r["y"])
            curves.append(np.pad(
                curve, (0, budget - len(curve)), mode="edge"
            ))
        curves = np.stack(curves)

        median = np.median(curves, axis=0)
        q25, q75 = np.quantile(curves, [0.25, 0.75], axis=0)

        ax.plot(
            evaluation_counts, median,
            label=name, color=colours[name],
        )
        ax.fill_between(
            evaluation_counts, q25, q75,
            color=colours[name], alpha=0.15,
        )

    ax.axhline(1.0, color="black", linestyle="--", linewidth=1)
    ax.set_yscale("symlog", linthresh=1e-4)
    ax.set_title(f"c = {c_value:g}")
    ax.set_xlabel("Objective evaluations")
    ax.set_ylabel("Best-so-far value")
    ax.grid(alpha=0.3)

axes[0, 0].legend()
fig.suptitle("Median progress; shading = 25th–75th percentiles")
plt.tight_layout()
plt.show()
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display

# Load the existing experiment data.
saved = json.loads(Path("experiment1_sweep.json").read_text())
runs = saved["runs"]
cfg = saved["config"]

a = np.asarray(cfg["a"], dtype=float)
b = np.asarray(cfg["b"], dtype=float)
r0 = cfg["r0"]
bounds = np.asarray(cfg["bounds"], dtype=float)
methods = cfg["methods"]

# Diagnostic definition: "near a" means distance <= 1.
# This fixed radius is a descriptive choice, not a basin boundary.
near_a_radius = 1.0
last_n = 20

# ----------------------------------------------------------
# 1. Measure concentration using ALL recorded runs.
# ----------------------------------------------------------
rows = []

for run in runs:
    X = np.asarray(run["X"], dtype=float)
    y = np.asarray(run["y"], dtype=float)

    distance_a = np.linalg.norm(X - a, axis=1)
    distance_b = np.linalg.norm(X - b, axis=1)

    # The better bowl is active wherever it is lower than the worse bowl.
    qa = 1.0 + distance_a**2
    qb = (
        distance_b**2
        + run["c"] * np.maximum(0.0, distance_b - r0)**2
    )
    better_bowl_active = qb < qa

    hits = np.flatnonzero(y < 1.0)

    rows.append({
        "c": run["c"],
        "method": run["method"],
        "seed": run["seed"],
        "success": bool(len(hits)),
        "final_best": float(y.min()),
        "last20_near_a": np.mean(distance_a[-last_n:] <= near_a_radius),
        "last20_in_better_bowl": np.mean(better_bowl_active[-last_n:]),
        "ever_in_better_bowl": bool(better_bowl_active.any()),
        "first_hit": int(hits[0] + 1) if len(hits) else np.nan,
    })

diagnostics = pd.DataFrame(rows)

concentration_summary = (
    diagnostics.groupby(["c", "method", "success"], sort=False)
    .agg(
        runs=("seed", "size"),
        mean_last20_near_a=("last20_near_a", "mean"),
        mean_last20_in_better_bowl=("last20_in_better_bowl", "mean"),
        fraction_ever_in_better_bowl=("ever_in_better_bowl", "mean"),
    )
    .reset_index()
)

print("Concentration summary: proportions range from 0 to 1")
display(concentration_summary.round(3))

# Plot concentration separately for successful and failed runs.
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharey=True)

for ax, success in zip(axes, [False, True]):
    subset = diagnostics[diagnostics["success"] == success]

    for name in methods:
        part = (
            subset[subset["method"] == name]
            .groupby("c")["last20_near_a"]
            .mean()
        )
        ax.plot(part.index, part.values, marker="o", label=name)

    ax.set_xscale("symlog", linthresh=1)
    ax.set_xticks(cfg["c_values"])
    ax.set_xticklabels([f"{c:g}" for c in cfg["c_values"]])
    ax.set_ylim(-0.05, 1.05)
    ax.set_xlabel("Penalty strength c")
    ax.set_title("Successful runs" if success else "Failed runs")
    ax.grid(alpha=0.3)

axes[0].set_ylabel("Mean fraction of final 20 evaluations within distance 1 of a")
axes[1].legend()
plt.tight_layout()
plt.show()

# ----------------------------------------------------------
# 2. Example search trajectories at c = 30.
#    Choose the lowest seed in each success/failure group.
# ----------------------------------------------------------
c_to_plot = 30.0

x1 = np.linspace(bounds[0, 0], bounds[0, 1], 250)
x2 = np.linspace(bounds[1, 0], bounds[1, 1], 250)
G1, G2 = np.meshgrid(x1, x2)

DA = np.sqrt((G1 - a[0])**2 + (G2 - a[1])**2)
DB = np.sqrt((G1 - b[0])**2 + (G2 - b[1])**2)

QA = 1.0 + DA**2
QB = DB**2 + c_to_plot * np.maximum(0.0, DB - r0)**2
F = np.minimum(QA, QB)

fig, axes = plt.subplots(
    2, len(methods),
    figsize=(4 * len(methods), 8),
    sharex=True, sharey=True,
    constrained_layout=True,
)

selected_rows = []

for col, name in enumerate(methods):
    for row, success in enumerate([False, True]):
        ax = axes[row, col]

        ax.contour(
            G1, G2, F,
            levels=[0.05, 0.25, 0.5, 1, 2, 4, 8, 16, 32],
            colors="lightgray",
            linewidths=0.7,
        )

        # Red dashed boundary encloses the success region f < 1.
        ax.contour(
            G1, G2, F, levels=[1.0],
            colors="red", linestyles="--", linewidths=1.2,
        )

        ax.scatter(*a, marker="X", color="black", s=70)
        ax.scatter(*b, marker="*", color="red", s=130)
        ax.text(a[0], a[1] + 0.25, "a: worse", ha="center")
        ax.text(b[0], b[1] + 0.25, "b: best", ha="center")

        candidates = sorted(
            [
                r for r in runs
                if r["c"] == c_to_plot
                and r["method"] == name
                and bool(r["success"]) == success
            ],
            key=lambda r: r["seed"],
        )

        status = "success" if success else "failure"

        if not candidates:
            ax.set_title(f"{name}: no {status} runs")
        else:
            run = candidates[0]
            X = np.asarray(run["X"])
            y = np.asarray(run["y"])

            # Colour shows evaluation order; points are not connected
            # because successive evaluations need not form a path.
            ax.scatter(
                X[:, 0], X[:, 1],
                c=np.arange(1, len(X) + 1),
                cmap="viridis", vmin=1, vmax=cfg["budget"],
                s=28, alpha=0.85,
            )

            # Circle the best evaluated point.
            best = X[np.argmin(y)]
            ax.scatter(
                *best, s=130,
                facecolors="none", edgecolors="black",
                linewidths=1.5,
            )

            ax.set_title(
                f"{name}: {status}, seed={run['seed']}\n"
                f"best={y.min():.4g}"
            )

            selected_rows.append({
                "method": name,
                "status": status,
                "seed": run["seed"],
                "best": float(y.min()),
            })

        ax.set_aspect("equal")
        ax.set_xlim(bounds[0])
        ax.set_ylim(bounds[1])
        ax.set_xlabel("x1")
        if col == 0:
            ax.set_ylabel("x2")

colour_scale = plt.cm.ScalarMappable(
    norm=plt.Normalize(1, cfg["budget"]),
    cmap="viridis",
)
fig.colorbar(
    colour_scale, ax=axes.ravel().tolist(),
    label="Evaluation number: purple = early, yellow = late",
    shrink=0.8,
)
fig.suptitle(f"Search locations at c = {c_to_plot:g}")
plt.show()

display(pd.DataFrame(selected_rows))
```

```python
import cma
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Reuse these from your notebook. lower and upper are in objective coordinates.
lower = np.asarray(lower, dtype=float)
upper = np.asarray(upper, dtype=float)
a = np.asarray(a, dtype=float)
b = np.asarray(b, dtype=float)
width = upper - lower

# Replace with the exact seed list from your existing sweep if different.
seeds = range(1, 21)
checkpoints = (50, 200)
c_values = (3, 30)

# Copy any non-default CMA-ES options from your existing runner here.
# If it explicitly sets popsize, include that too.
baseline_options = {"bounds": [0.0, 1.0], "verbose": -9}
sigma0 = 0.30

a_unit = (a - lower) / width
b_unit = (b - lower) / width

# Matched starting points on the line between the minima.
# For example, 0.10 starts 10% of the way toward the other minimum.
start_fractions = (0.05, 0.10, 0.15)
start_points = [
    (basin, fraction, centre + fraction * (other - centre))
    for basin, centre, other in (
        ("a", a_unit, b_unit),
        ("b", b_unit, a_unit),
    )
    for fraction in start_fractions
]

probe = cma.CMAEvolutionStrategy(
    start_points[0][2], sigma0, {**baseline_options, "seed": 1}
)
baseline_popsize = probe.popsize
print(f"Population sizes: {baseline_popsize} and {2 * baseline_popsize}")


def run_population_test(c, start_basin, start_fraction, x0, seed, popsize):
    es = cma.CMAEvolutionStrategy(
        x0,
        sigma0,
        {**baseline_options, "seed": int(seed), "popsize": popsize},
    )
    evaluations = 0
    last_update = 0
    best_f = np.inf
    first_hit = None
    records = []

    def record_state(checkpoint):
        # Distances are in the same [0, 1]^2 coordinates that CMA-ES uses.
        # Include pycma's coordinate scaling if present.
        mean = np.asarray(es.mean)
        scaling = np.asarray(es.sigma_vec.scaling)
        covariance = es.sigma**2 * np.asarray(es.C) * np.outer(scaling, scaling)

        mean_dist_a = np.linalg.norm(mean - a_unit)
        mean_dist_b = np.linalg.norm(mean - b_unit)
        variance = np.trace(covariance)

        records.append({
            "c": c,
            "start_basin": start_basin,
            "start_fraction": start_fraction,
            "seed": seed,
            "popsize": popsize,
            "evaluations": evaluations,
            "last_update": last_update,
            "checkpoint": checkpoint,
            "D_a": np.sqrt(mean_dist_a**2 + variance),
            "D_b": np.sqrt(mean_dist_b**2 + variance),
            "mean_dist_a": mean_dist_a,
            "mean_dist_b": mean_dist_b,
            "spread": np.sqrt(variance),
            "best_f": best_f,
            "first_hit": first_hit,
        })

    record_state(checkpoint=False)

    while evaluations < max(checkpoints):
        candidates = es.ask()
        values = []

        for candidate in candidates:
            x = lower + width * np.asarray(candidate)
            value = float(two_basin_function(x, c=c, a=a, b=b, r0=r0))

            values.append(value)
            evaluations += 1
            best_f = min(best_f, value)
            if first_hit is None and value < 1:
                first_hit = evaluations

            # Only completed generations update the sampling distribution.
            completed_generation = len(values) == len(candidates)
            if completed_generation:
                es.tell(candidates, values)
                last_update = evaluations

            at_checkpoint = evaluations in checkpoints
            if completed_generation or at_checkpoint:
                record_state(checkpoint=at_checkpoint)

            if evaluations == max(checkpoints):
                break

    return records


history = pd.DataFrame([
    row
    for c in c_values
    for start_basin, start_fraction, x0 in start_points
    for seed in seeds
    for popsize in (baseline_popsize, 2 * baseline_popsize)
    for row in run_population_test(
        c, start_basin, start_fraction, x0, seed, popsize
    )
])
```

```python
from itertools import product

import cma
import numpy as np
import pandas as pd

# Existing notebook variables: a, b, r0, lower, upper, two_basin_function.
experiment_configs = list(product((6, 12), (3, 6)))
experiment_budgets = (50, 200)
experiment_c_values = (3, 30)
experiment_seeds = range(1, 21)
experiment_fractions = (0.05, 0.10, 0.15)
experiment_sigma0 = 0.30

domain_lower = np.asarray(lower, dtype=float)
domain_width = np.asarray(upper, dtype=float) - domain_lower
a_unit = (np.asarray(a) - domain_lower) / domain_width
b_unit = (np.asarray(b) - domain_lower) / domain_width

experiment_starts = [
    (basin, fraction, centre + fraction * (other - centre))
    for basin, centre, other in (
        ("a", a_unit, b_unit),
        ("b", b_unit, a_unit),
    )
    for fraction in experiment_fractions
]


def make_cma_strategy(x0, seed, popsize, mu):
    es = cma.CMAEvolutionStrategy(
        x0,
        experiment_sigma0,
        {
            "bounds": [0.0, 1.0],
            "seed": seed,
            "popsize": popsize,
            "CMA_mu": mu,
            "verbose": -9,
            "verb_log": 0,
        },
    )
    actual = (es.popsize, es.sp.weights.mu)
    if actual != (popsize, mu):
        raise ValueError(f"Requested {(popsize, mu)}, but pycma used {actual}")
    return es


for popsize, mu in experiment_configs:
    probe = make_cma_strategy(experiment_starts[0][2], 1, popsize, mu)
    print(
        f"λ={probe.popsize}, μ={probe.sp.weights.mu}, "
        f"mean weights={np.round(probe.sp.weights[:mu], 4)}"
    )


def run_cma_configuration(c_value, basin, fraction, x0, seed, popsize, mu):
    es = make_cma_strategy(x0, seed, popsize, mu)
    evaluations = 0
    last_update = 0
    best_f = np.inf
    first_hit = None
    records = []

    def record_state():
        mean = np.asarray(es.mean)
        scaling = np.broadcast_to(np.asarray(es.sigma_vec.scaling), mean.shape)
        variance = es.sigma**2 * np.dot(np.diag(es.C), scaling**2)
        mean_dist_a = np.linalg.norm(mean - a_unit)
        mean_dist_b = np.linalg.norm(mean - b_unit)

        records.append({
            "c": c_value,
            "start_basin": basin,
            "start_fraction": fraction,
            "seed": seed,
            "popsize": popsize,
            "mu": mu,
            "evaluations": evaluations,
            "last_update": last_update,
            "checkpoint": evaluations in experiment_budgets,
            "D_a": np.sqrt(mean_dist_a**2 + variance),
            "D_b": np.sqrt(mean_dist_b**2 + variance),
            "mean_dist_a": mean_dist_a,
            "mean_dist_b": mean_dist_b,
            "spread": np.sqrt(variance),
            "best_f": best_f,
            "first_hit": first_hit,
        })

    record_state()
    max_budget = max(experiment_budgets)

    while evaluations < max_budget:
        candidates = es.ask()
        values = []

        for candidate in candidates:
            x = domain_lower + domain_width * np.asarray(candidate)
            value = float(two_basin_function(x, c_value, a, b, r0))
            values.append(value)
            evaluations += 1
            best_f = min(best_f, value)

            if first_hit is None and value < 1.0:
                first_hit = evaluations

            completed_generation = len(values) == len(candidates)
            if completed_generation:
                es.tell(candidates, values)
                last_update = evaluations

            if completed_generation or evaluations in experiment_budgets:
                record_state()

            if evaluations == max_budget:
                break

    return records


factorial_rows = []

for popsize, mu in experiment_configs:
    for c_value in experiment_c_values:
        for basin, fraction, x0 in experiment_starts:
            for seed in experiment_seeds:
                factorial_rows.extend(
                    run_cma_configuration(
                        c_value, basin, fraction, x0, seed, popsize, mu
                    )
                )
    print(f"Finished λ={popsize}, μ={mu}")

factorial_history = pd.DataFrame(factorial_rows)
```

```python
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

import cma
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

LOWER = np.array([-20.0, -20.0])
UPPER = np.array([20.0, 20.0])
WIDTH = UPPER - LOWER

A = np.array([-10.0, 0.0])
B = np.array([10.0, 0.0])
R0 = 0.5

A_UNIT = (A - LOWER) / WIDTH
B_UNIT = (B - LOWER) / WIDTH

SIGMA0 = 0.30
C_VALUES = (3, 30)
BUDGETS = (50, 200)
SEEDS = range(1, 21)
START_FRACTIONS = (0.05, 0.10, 0.15)

LAMBDA_VALUES = (6, 8, 12, 16, 24, 32, 48)
MU_VALUES = (2, 4, 6, 8, 10)
LAMBDA_CONFIGS = [(lam, lam // 2) for lam in LAMBDA_VALUES]
MU_CONFIGS = [(12, mu) for mu in MU_VALUES]
CONFIGS = list(dict.fromkeys(LAMBDA_CONFIGS + MU_CONFIGS))

# Same physical concentration cutoff as before.
# Original: 0.05 × 8 = 0.40. New: 0.01 × 40 = 0.40.
NEAR_B_CUTOFF = 0.40 / WIDTH[0]

STARTS = [
    (basin, fraction, centre + fraction * (other - centre))
    for basin, centre, other in (
        ("a", A_UNIT, B_UNIT),
        ("b", B_UNIT, A_UNIT),
    )
    for fraction in START_FRACTIONS
]

display(pd.DataFrame([
    {
        "start_basin": basin,
        "fraction_toward_other_minimum": fraction,
        "start_x": (LOWER + WIDTH * x0)[0],
        "start_y": (LOWER + WIDTH * x0)[1],
    }
    for basin, fraction, x0 in STARTS
]))


def objective(x, c_value):
    distance_a = np.linalg.norm(x - A)
    distance_b = np.linalg.norm(x - B)
    bowl_a = 1.0 + distance_a**2
    bowl_b = distance_b**2 + c_value * max(0.0, distance_b - R0)**2
    return float(min(bowl_a, bowl_b))


assert objective(A, 30) == 1.0
assert objective(B, 30) == 0.0
```

```python
def make_strategy(x0, seed, lam, mu):
    es = cma.CMAEvolutionStrategy(
        x0.copy(),
        SIGMA0,
        {
            "bounds": [0.0, 1.0],
            "seed": int(seed),
            "popsize": lam,
            "CMA_mu": mu,
            "verbose": -9,
            "verb_log": 0,
        },
    )

    actual = (es.popsize, es.sp.weights.mu)
    if actual != (lam, mu):
        raise ValueError(f"Requested {(lam, mu)}, but pycma used {actual}")
    return es


def run_wide_cma(c_value, basin, fraction, x0, seed, lam, mu):
    es = make_strategy(x0, seed, lam, mu)
    max_budget = max(BUDGETS)

    evaluations = 0
    last_update = 0
    best_f = np.inf
    first_hit_eval = None
    first_hit_f = None
    first_hit_dist_b = None
    first_refined_eval = None
    records = []

    def record_state():
        mean = np.asarray(es.mean)
        scaling = np.broadcast_to(np.asarray(es.sigma_vec.scaling), mean.shape)
        variance = es.sigma**2 * np.dot(np.diag(es.C), scaling**2)
        mean_dist_a = np.linalg.norm(mean - A_UNIT)
        mean_dist_b = np.linalg.norm(mean - B_UNIT)

        records.append({
            "c": c_value,
            "start_basin": basin,
            "start_fraction": fraction,
            "seed": seed,
            "popsize": lam,
            "mu": mu,
            "evaluations": evaluations,
            "last_update": last_update,
            "checkpoint": evaluations in BUDGETS,
            "D_a": np.sqrt(mean_dist_a**2 + variance),
            "D_b": np.sqrt(mean_dist_b**2 + variance),
            "mean_dist_a": mean_dist_a,
            "mean_dist_b": mean_dist_b,
            "spread": np.sqrt(variance),
            "best_f": best_f,
            "first_hit_eval": first_hit_eval,
            "first_hit_f": first_hit_f,
            "first_hit_dist_b": first_hit_dist_b,
            "first_refined_eval": first_refined_eval,
        })

    record_state()

    while evaluations < max_budget:
        candidates = es.ask()
        values = []

        for candidate in candidates:
            x = LOWER + WIDTH * np.asarray(candidate)
            value = objective(x, c_value)
            values.append(value)
            evaluations += 1
            best_f = min(best_f, value)

            if first_hit_eval is None and value < 1.0:
                first_hit_eval = evaluations
                first_hit_f = value
                first_hit_dist_b = float(np.linalg.norm(x - B))

            if first_refined_eval is None and value < 0.01:
                first_refined_eval = evaluations

            completed_generation = len(values) == len(candidates)
            if completed_generation:
                es.tell(candidates, values)
                last_update = evaluations

            if completed_generation or evaluations in BUDGETS:
                record_state()

            if evaluations == max_budget:
                break

    return records


# Verify every configuration before running the sweep.
for lam, mu in CONFIGS:
    make_strategy(STARTS[0][2], 1, lam, mu)

wide_rows = []
for lam, mu in CONFIGS:
    for c_value in C_VALUES:
        for basin, fraction, x0 in STARTS:
            for seed in SEEDS:
                wide_rows.extend(
                    run_wide_cma(c_value, basin, fraction, x0, seed, lam, mu)
                )
    print(f"Finished λ={lam}, μ={mu}")

wide_history = pd.DataFrame(wide_rows)
wide_checkpoints = wide_history.loc[wide_history["checkpoint"]].copy()

expected_runs = len(CONFIGS) * len(C_VALUES) * len(STARTS) * len(SEEDS)
assert len(wide_checkpoints) == expected_runs * len(BUDGETS)
print(f"Completed {expected_runs:,} runs")
```

```python
wide_checkpoints["discovered"] = wide_checkpoints["best_f"].lt(1.0)
wide_checkpoints["refined"] = wide_checkpoints["best_f"].lt(0.01)
wide_checkpoints["near_b"] = wide_checkpoints["D_b"].le(NEAR_B_CUTOFF)

wide_summary = (
    wide_checkpoints.groupby(
        ["c", "start_basin", "evaluations", "popsize", "mu"],
        as_index=False,
    )
    .agg(
        runs=("seed", "size"),
        discoveries=("discovered", "sum"),
        refinements=("refined", "sum"),
        finished_near_b=("near_b", "sum"),
        median_best_f=("best_f", "median"),
    )
)

for metric in ("discoveries", "refinements", "finished_near_b"):
    wide_summary[f"{metric}_pct"] = 100 * wide_summary[metric] / wide_summary["runs"]


def split_sweeps(table):
    return [
        (
            table.loc[table["mu"].eq(table["popsize"] // 2)],
            "popsize", LAMBDA_VALUES, "Population λ (μ = λ/2)",
        ),
        (
            table.loc[table["popsize"].eq(12) & table["mu"].isin(MU_VALUES)],
            "mu", MU_VALUES, "Parents μ (λ = 12)",
        ),
    ]


wide_final = wide_summary.loc[wide_summary["evaluations"].eq(200)]

best_tables = []
for table, _, _, label in split_sweeps(wide_final):
    for metric in ("refinements", "finished_near_b"):
        peak = table.groupby(["c", "start_basin"])[metric].transform("max")
        best_tables.append(
            table.loc[table[metric].eq(peak)].assign(sweep=label, metric=metric)
        )

wide_best_settings = pd.concat(best_tables, ignore_index=True)

display(wide_best_settings[[
    "sweep", "metric", "c", "start_basin", "popsize", "mu",
    "discoveries", "refinements", "finished_near_b", "runs",
]])
```

```python
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

lam, mu = 12, 6
initial_sds = (2.4, 12.0)
offsets = (0.2, 0.4, 0.6)

# Convert physical offsets to fractions along the line between A and B.
separation = np.linalg.norm(B - A)
sigma_starts = [
    (basin, offset, centre + (offset / separation) * (other - centre))
    for basin, centre, other in (
        ("a", A_UNIT, B_UNIT),
        ("b", B_UNIT, A_UNIT),
    )
    for offset in offsets
]

sigma_rows = []
original_sigma0 = SIGMA0

try:
    for initial_sd in initial_sds:
        SIGMA0 = initial_sd / WIDTH[0]

        for c_value in C_VALUES:
            for basin, offset, x0 in sigma_starts:
                for seed in SEEDS:
                    records = run_wide_cma(
                        c_value, basin, offset / separation, x0, seed, lam, mu
                    )
                    for record in records:
                        record["initial_sd"] = initial_sd
                        record["start_offset"] = offset
                    sigma_rows.extend(records)

        print(f"Finished initial SD={initial_sd:g} (SIGMA0={SIGMA0:g})")
finally:
    SIGMA0 = original_sigma0

sigma_history = pd.DataFrame(sigma_rows)
sigma_checkpoints = sigma_history.loc[sigma_history["checkpoint"]].copy()

sigma_checkpoints["first_generation_hit"] = (
    sigma_checkpoints["first_hit_eval"].le(lam)
)
sigma_checkpoints["discovered"] = sigma_checkpoints["best_f"].lt(1.0)
sigma_checkpoints["refined"] = sigma_checkpoints["best_f"].lt(0.01)
sigma_checkpoints["near_b"] = sigma_checkpoints["D_b"].le(NEAR_B_CUTOFF)

sigma_summary = (
    sigma_checkpoints
    .groupby(["c", "start_basin", "evaluations", "initial_sd"], as_index=False)
    .agg(
        runs=("seed", "size"),
        first_generation_hits=("first_generation_hit", "sum"),
        discoveries=("discovered", "sum"),
        refinements=("refined", "sum"),
        finished_near_b=("near_b", "sum"),
        median_best_f=("best_f", "median"),
    )
)

display(sigma_summary.round(5))
```

```python
from itertools import product
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

import cma
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display

LOWER = -10.0
UPPER = 10.0
WIDTH = UPPER - LOWER

A = np.array([-5.0, 0.0])
B = np.array([5.0, 0.0])
R0 = 0.5

A_UNIT = (A - LOWER) / WIDTH
B_UNIT = (B - LOWER) / WIDTH

INITIAL_SD = 6.0                  # Physical units. Set to 2.4 for a narrower start.
SIGMA0 = INITIAL_SD / WIDTH       # pycma uses normalised coordinates.
NEAR_B_CUTOFF = 0.40 / WIDTH     # Same physical RMS cutoff as before.

C_VALUES = (3, 30)
BUDGETS = (50, 200)
SEEDS = range(1, 21)
START_FRACTIONS = (0.05, 0.10, 0.15)

LAMBDA_VALUES = (6, 8, 12, 16, 24, 32, 48)
MU_VALUES = (2, 4, 6, 8, 10)
LAMBDA_CONFIGS = [(lam, lam // 2) for lam in LAMBDA_VALUES]
MU_CONFIGS = [(12, mu) for mu in MU_VALUES]
CONFIGS = list(dict.fromkeys(LAMBDA_CONFIGS + MU_CONFIGS))

STARTS = [
    (basin, fraction, centre + fraction * (other - centre))
    for basin, centre, other in (
        ("a", A_UNIT, B_UNIT),
        ("b", B_UNIT, A_UNIT),
    )
    for fraction in START_FRACTIONS
]


def objective(x, c_value):
    distance_a = np.linalg.norm(x - A)
    distance_b = np.linalg.norm(x - B)
    bowl_a = 1.0 + distance_a**2
    bowl_b = distance_b**2 + c_value * max(0.0, distance_b - R0)**2
    return float(min(bowl_a, bowl_b))


assert objective(A, 30) == 1.0
assert objective(B, 30) == 0.0

display(pd.DataFrame([
    {
        "start_basin": basin.upper(),
        "start_x": (LOWER + WIDTH * x0)[0],
        "start_y": (LOWER + WIDTH * x0)[1],
        "distance_from_starting_minimum": fraction * np.linalg.norm(B - A),
    }
    for basin, fraction, x0 in STARTS
]).round(3))
```

```python
def make_strategy(x0, seed, lam, mu):
    es = cma.CMAEvolutionStrategy(
        x0.copy(),
        SIGMA0,
        {
            "bounds": [0.0, 1.0],
            "seed": int(seed),
            "popsize": lam,
            "CMA_mu": mu,
            "verbose": -9,
            "verb_log": 0,
        },
    )
    actual = (es.popsize, es.sp.weights.mu)
    if actual != (lam, mu):
        raise ValueError(f"Requested {(lam, mu)}, but pycma used {actual}")
    return es


def run_cma(c_value, basin, fraction, x0, seed, lam, mu):
    es = make_strategy(x0, seed, lam, mu)
    max_budget = max(BUDGETS)
    records = []

    evaluations = 0
    last_update = 0
    best_f = np.inf
    first_hit_eval = None
    first_hit_f = None
    first_hit_dist_b = None
    first_refined_eval = None

    def record_state():
        mean = np.asarray(es.mean)
        scaling = np.broadcast_to(np.asarray(es.sigma_vec.scaling), mean.shape)
        variance = es.sigma**2 * np.dot(np.diag(es.C), scaling**2)
        mean_dist_a = np.linalg.norm(mean - A_UNIT)
        mean_dist_b = np.linalg.norm(mean - B_UNIT)

        records.append({
            "c": c_value,
            "start_basin": basin,
            "start_fraction": fraction,
            "seed": seed,
            "popsize": lam,
            "mu": mu,
            "evaluations": evaluations,
            "last_update": last_update,
            "checkpoint": evaluations in BUDGETS,
            "D_a": np.sqrt(mean_dist_a**2 + variance),
            "D_b": np.sqrt(mean_dist_b**2 + variance),
            "spread": np.sqrt(variance),
            "best_f": best_f,
            "first_hit_eval": first_hit_eval,
            "first_hit_f": first_hit_f,
            "first_hit_dist_b": first_hit_dist_b,
            "first_refined_eval": first_refined_eval,
        })

    record_state()

    while evaluations < max_budget:
        candidates = es.ask()
        values = []

        for candidate in candidates:
            x = LOWER + WIDTH * np.asarray(candidate)
            value = objective(x, c_value)
            values.append(value)
            evaluations += 1
            best_f = min(best_f, value)

            if first_hit_eval is None and value < 1.0:
                first_hit_eval = evaluations
                first_hit_f = value
                first_hit_dist_b = float(np.linalg.norm(x - B))

            if first_refined_eval is None and value < 0.01:
                first_refined_eval = evaluations

            completed_generation = len(values) == len(candidates)
            if completed_generation:
                es.tell(candidates, values)
                last_update = evaluations

            if completed_generation or evaluations in BUDGETS:
                record_state()

            if evaluations == max_budget:
                break

    return records


# Check the effective population and parent counts before the full sweep.
for lam, mu in CONFIGS:
    make_strategy(STARTS[0][2], 1, lam, mu)

rows = []
for lam, mu in CONFIGS:
    for c_value, (basin, fraction, x0), seed in product(C_VALUES, STARTS, SEEDS):
        rows.extend(run_cma(c_value, basin, fraction, x0, seed, lam, mu))
    print(f"Finished λ={lam}, μ={mu}")

mid_history = pd.DataFrame(rows)
mid_checkpoints = mid_history.loc[mid_history["checkpoint"]].copy()

expected_runs = len(CONFIGS) * len(C_VALUES) * len(STARTS) * len(SEEDS)
assert len(mid_checkpoints) == expected_runs * len(BUDGETS)
print(f"Completed {expected_runs:,} runs")
```

## mean learning rate & Step-size adaptatio

```python
from itertools import product

import cma
import numpy as np
import pandas as pd
from IPython.display import display

LOWER, UPPER = -4.0, 4.0
WIDTH = UPPER - LOWER

A = np.array([-2.0, 0.0])
B = np.array([2.0, 0.0])
R0 = 0.5

A_UNIT = (A - LOWER) / WIDTH
B_UNIT = (B - LOWER) / WIDTH

POP_SIZE, PARENT_COUNT = 12, 6
SIGMA0 = 0.30                        # Normalised. Initial physical SD = 2.4.
C_VALUES = (3, 30)
BUDGETS = (50, 200)
SEEDS = range(1, 21)
START_FRACTIONS = (0.05, 0.10, 0.15)
NEAR_CUTOFF = 0.40                    # Physical RMS distance, as before.

VARIANTS = {
    "Baseline": {"CMA_cmean": 1.0, "AdaptSigma": True},
    "Half mean step": {"CMA_cmean": 0.5, "AdaptSigma": True},
    "Fixed step size": {"CMA_cmean": 1.0, "AdaptSigma": False},
}

STARTS = [
    (basin, fraction, centre + fraction * (other - centre))
    for basin, centre, other in (
        ("A", A_UNIT, B_UNIT),
        ("B", B_UNIT, A_UNIT),
    )
    for fraction in START_FRACTIONS
]


def objective(x, c_value):
    dist_a = np.linalg.norm(x - A)
    dist_b = np.linalg.norm(x - B)
    bowl_a = 1.0 + dist_a**2
    bowl_b = dist_b**2 + c_value * max(0.0, dist_b - R0)**2
    return min(bowl_a, bowl_b)


def run_update_test(c_value, basin, fraction, x0, seed, variant):
    options = VARIANTS[variant]
    es = cma.CMAEvolutionStrategy(
        x0.copy(),
        SIGMA0,
        {
            "bounds": [0.0, 1.0],
            "seed": seed,
            "popsize": POP_SIZE,
            "CMA_mu": PARENT_COUNT,
            "verbose": -9,
            "verb_log": 0,
            **options,
        },
    )
    assert (es.popsize, es.sp.weights.mu) == (POP_SIZE, PARENT_COUNT)

    run_id = {
        "c": c_value,
        "start_basin": basin,
        "start_fraction": fraction,
        "seed": seed,
        "variant": variant,
    }
    evaluations = 0
    best_f = np.inf
    first_hit = None
    first_refined = None
    first_hit_dist_b = np.nan
    mean_gain_on_hit_update = np.nan
    rows = []
    max_budget = max(BUDGETS)

    def state_metrics():
        mean = np.asarray(es.mean)
        scaling = np.broadcast_to(np.asarray(es.sigma_vec.scaling), mean.shape)
        variance = (WIDTH * es.sigma)**2 * np.dot(np.diag(es.C), scaling**2)
        dist_a = WIDTH * np.linalg.norm(mean - A_UNIT)
        dist_b = WIDTH * np.linalg.norm(mean - B_UNIT)

        return {
            "mean_dist_a": dist_a,
            "mean_dist_b": dist_b,
            "sigma": WIDTH * es.sigma,
            "spread": np.sqrt(variance),
            "D_a": np.sqrt(dist_a**2 + variance),
            "D_b": np.sqrt(dist_b**2 + variance),
        }

    state = state_metrics()

    while evaluations < max_budget:
        candidates = es.ask()
        values = []
        first_hit_this_generation = False

        for candidate in candidates:
            x = LOWER + WIDTH * np.asarray(candidate)
            value = float(objective(x, c_value))
            values.append(value)
            evaluations += 1
            best_f = min(best_f, value)

            if first_hit is None and value < 1.0:
                first_hit = evaluations
                first_hit_dist_b = np.linalg.norm(x - B)
                first_hit_this_generation = True

            if first_refined is None and value < 0.01:
                first_refined = evaluations

            if len(values) == len(candidates):
                previous_dist_b = state["mean_dist_b"]
                es.tell(candidates, values)
                state = state_metrics()

                if first_hit_this_generation:
                    mean_gain_on_hit_update = (
                        previous_dist_b - state["mean_dist_b"]
                    )

                if variant == "Fixed step size":
                    assert np.isclose(es.sigma, SIGMA0)

            rows.append({
                **run_id,
                "evaluations": evaluations,
                **state,
                "best_f": best_f,
                "first_hit": first_hit,
                "first_refined": first_refined,
                "first_hit_dist_b": first_hit_dist_b,
                "mean_gain_on_hit_update": mean_gain_on_hit_update,
            })

            if evaluations == max_budget:
                break

    return rows


rows = []
for variant in VARIANTS:
    for c_value, (basin, fraction, x0), seed in product(C_VALUES, STARTS, SEEDS):
        rows.extend(run_update_test(c_value, basin, fraction, x0, seed, variant))
    print(f"Finished {variant}")

history = pd.DataFrame(rows)
history["discovered"] = history["best_f"].lt(1.0)
history["refined"] = history["best_f"].lt(0.01)
history["near_a"] = history["D_a"].le(NEAR_CUTOFF)
history["near_b"] = history["D_b"].le(NEAR_CUTOFF)

checkpoints = history.loc[history["evaluations"].isin(BUDGETS)].copy()
summary = (
    checkpoints.groupby(["c", "start_basin", "evaluations", "variant"], sort=False)
    .agg(
        runs=("seed", "size"),
        discovered=("discovered", "sum"),
        refined=("refined", "sum"),
        near_a=("near_a", "sum"),
        near_b=("near_b", "sum"),
        median_best_f=("best_f", "median"),
        median_sigma=("sigma", "median"),
        median_spread=("spread", "median"),
        hits_with_completed_update=("mean_gain_on_hit_update", "count"),
        median_mean_gain_after_hit=("mean_gain_on_hit_update", "median"),
    )
    .reset_index()
)

assert summary["runs"].eq(len(SEEDS) * len(START_FRACTIONS)).all()
display(summary.round(5))
```
